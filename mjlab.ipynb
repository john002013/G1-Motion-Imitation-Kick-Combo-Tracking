```python
import subprocess, os

env = os.environ.copy()
env['MUJOCO_GL']     = 'egl'
env['WANDB_ENTITY']  = 'abidemibabalola23-ladoke-akintola-university-of-technology'
env['WANDB_API_KEY'] = wandb_key

# nohup makes it survive cell interruptions
with open('/kaggle/working/tracking_train.log', 'w') as log:
    process = subprocess.Popen(
        [
            'nohup',
            '/kaggle/working/mjlab/.venv/bin/python', '-m', 'mjlab.scripts.train',
            'Mjlab-Tracking-Flat-Unitree-G1',
            '--registry-name',
            'abidemibabalola23-ladoke-akintola-university-of-technology/csv_to_npz/g1_kick:v0',
            '--env.scene.num-envs', '1024',
            '--agent.max-iterations', '3000',
            '--agent.save-interval', '50',
        ],
        env=env,
        cwd='/kaggle/working/mjlab',
        stdout=log,
        stderr=log,
        start_new_session=True,  # ← this is the key, detaches from notebook
    )

# Save PID so we can check it later
with open('/kaggle/working/train.pid', 'w') as f:
    f.write(str(process.pid))

print(f"Training started with PID: {process.pid}")
print("Safe to stop any cell now — training will keep running")
```

**Resuming**#   

```python
import subprocess, os
from kaggle_secrets import UserSecretsClient

secrets = UserSecretsClient()
wandb_key = secrets.get_secret("WANDB_API_KEY")

env = os.environ.copy()
env['MUJOCO_GL']     = 'egl'
env['WANDB_ENTITY']  = 'abidemibabalola23-ladoke-akintola-university-of-technology'
env['WANDB_API_KEY'] = wandb_key

with open('/kaggle/working/tracking_train.log', 'w') as log:
    process = subprocess.Popen(
        [
            'uv', 'run', 'train',
            'Mjlab-Tracking-Flat-Unitree-G1',
            '--registry-name',
            'abidemibabalola23-ladoke-akintola-university-of-technology/csv_to_npz/g1_kick:v0',
            '--env.scene.num-envs', '1024',
            '--agent.max-iterations', '2000',  # ← 2000 more on top of 3000
            '--agent.save-interval', '50',
            '--agent.resume', 'True',           # ← resume from last checkpoint
            '--wandb-run-path',
            'abidemibabalola23-ladoke-akintola-university-of-technology/mjlab/ahc620qc',
        ],
        env=env,
        cwd='/kaggle/working/mjlab',
        stdout=log,
        stderr=log,
        start_new_session=True,
    )

with open('/kaggle/working/train.pid', 'w') as f:
    f.write(str(process.pid))

print(f"Resumed training! PID: {process.pid}")
```
